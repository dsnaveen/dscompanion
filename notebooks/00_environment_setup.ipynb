# Environment Setup (conda / Miniconda)

Before running any of the other notebooks, set up a dedicated Python environment for
`dscompanion`. This isn't strictly required — a plain `venv` works too — but a
dedicated conda environment is the most common setup for a data science workflow, and
it's what this notebook walks through.

**Why bother with a dedicated environment at all?** `dscompanion` pins specific
versions of `pandas`, `numpy`, `scikit-learn`, `xgboost`, and friends
(see [`pyproject.toml`](../pyproject.toml)). Installing it into a Python environment
you also use for other projects risks version conflicts — an isolated environment
avoids that entirely, and makes it trivial to remove everything cleanly later
(`conda env remove -n dscompanion`).

**Important — this notebook is a guide, not a script.** Creating and activating a
conda environment changes *your terminal's* shell state; it can't be done *from
inside* an already-running Jupyter kernel (running `!conda activate ...` as a
notebook cell only affects a short-lived subprocess, not the kernel you're reading
this in). So the setup commands below are meant to be **copied into your own
terminal**, not run as notebook cells. The executable cells here are just
verification checks, safe to run against whatever environment is powering this
kernel right now.

## Step 1: Do you already have conda?

Check in your terminal:

```bash
conda --version
```

If that prints a version number, skip to **Step 3**. If it says `command not found`,
continue to **Step 2**.

````python
# Safe to run from here: just checks whether conda is available to *this* kernel's
# underlying shell. A missing conda here doesn't mean much either way — what matters
# is whether it's on YOUR terminal's PATH, which this notebook can't see (e.g. this
# check will correctly report "not found" in a CI runner that never installed conda
# at all, which is expected and not a problem).
import subprocess

try:
    result = subprocess.run(["conda", "--version"], capture_output=True, text=True)
    print(result.stdout.strip() or "conda ran but printed nothing")
except FileNotFoundError:
    print("conda not found on this kernel's PATH")
````

## Step 2: Install Miniconda

Miniconda is the minimal conda installer (a few hundred MB) — plenty for this. Full
Anaconda works too if you already have it, but Miniconda is the lighter choice if
you're starting fresh.

**macOS (Apple Silicon):**
```bash
curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-MacOSX-arm64.sh
bash Miniconda3-latest-MacOSX-arm64.sh
```

**macOS (Intel):**
```bash
curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-MacOSX-x86_64.sh
bash Miniconda3-latest-MacOSX-x86_64.sh
```

**Linux (x86_64):**
```bash
curl -O https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
bash Miniconda3-latest-Linux-x86_64.sh
```

**Windows:** download and run the graphical installer from
[the official Miniconda page](https://docs.conda.io/en/latest/miniconda.html).

Follow the installer's prompts (accept the license, accept the default install
location), then **close and reopen your terminal** so the `conda` command is picked
up. Re-run `conda --version` from Step 1 to confirm it worked.

## Step 3: Create the environment

`dscompanion` requires Python 3.12+ (see `requires-python` in
[`pyproject.toml`](../pyproject.toml)), so pin the new environment to that:

```bash
conda create -n dscompanion python=3.12 -y
```

## Step 4: Activate it

```bash
conda activate dscompanion
```

Your terminal prompt should now show `(dscompanion)` at the start of the line.
**Every command from here on assumes you're inside this activated environment.**

## Step 5: Install dscompanion

```bash
pip install dscompanion[notebooks]
```

The `[notebooks]` extra pulls in `jupyter`, `ipykernel`, and `ucimlrepo` (the UCI
dataset fetch client) on top of `dscompanion` itself — everything the other notebooks
in this folder need.

## Step 6: Register this environment as a Jupyter kernel

If you're launching Jupyter from a different (e.g. base) environment and want to run
these notebooks against the `dscompanion` environment you just built, register it as
an explicit kernel:

```bash
python -m ipykernel install --user --name=dscompanion --display-name "Python (dscompanion)"
```

Then, in Jupyter's interface, use **Kernel → Change Kernel → Python (dscompanion)**
(or the equivalent kernel picker in your IDE) to switch this notebook — and the
others in this folder — onto it.

If you launched Jupyter *from inside* the activated `dscompanion` environment
directly (`conda activate dscompanion` then `jupyter notebook`), this step isn't
necessary — you're already running on the right environment.
